### Dataset:
The Iris dataset contains 150 instances with 4 features (sepal length, sepal width, petal length, and petal width) and a target label representing the species of Iris flowers (Setosa, Versicolour, Virginica).

### Steps
1. Load the Dataset

```python
import pandas as pd
import tensorflow_datasets as tfds

# Load the Iris dataset from TensorFlow Datasets
ds = tfds.load('iris', split='train', as_supervised=False)
df = tfds.as_dataframe(ds)

# Save the dataset as a CSV file
df.to_csv("iris_dataset.csv", index=False)

# Load the CSV file
iris_df = pd.read_csv("iris_dataset.csv")
iris_df.head()
```

### 2. Preprocess the Data
-  Handle Missing Values: Verify if the dataset contains missing values and handle them appropriately.
- Normalize Numerical Features: Scale numerical features for better training performance.
- Encode Categorical Variables: Convert the target label (species) to one-hot encoding.

```python
# Load the CSV file
df = pd.read_csv("iris_dataset.csv")

# Check for missing values and handle them (if any)
print(df.isnull().sum())  # Check for missing values
df = df.dropna()  # Drop rows with missing values if any

# Normalize the numerical features (sepal_length, sepal_width, petal_length, petal_width)
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

# Extracting numerical columns for normalization
numerical_features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
df[numerical_features] = scaler.fit_transform(df[numerical_features])

# One-hot encode the target column (species)
df = pd.get_dummies(df, columns=['species'], drop_first=True)

# Show the first few rows of the preprocessed data
print(df.head())
```
